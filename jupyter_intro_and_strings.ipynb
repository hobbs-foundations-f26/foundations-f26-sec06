Markdown is a markup language (like HTML: HypText Markup Language).

You do *not* need to know this for class. But especially in recent years (think generative AI), it has become a lot more widespread.

Markdown is *not* a programming language (you can not calculate 3+4).  It is a markup language.  That means it is text that is supposed to be rendered on a screen.  Rendering is drawing.  This is so that humans can look at pretty things on the screen.

Some examples of markdown.

# Big heading
## so-so big heading
### even smaller heading
#### whoa, how far can you go?

You can make lists:
1. one
2. two
3. and have *italics*

You can also make bullets:
- hey
- that's
- pretty 
- easy

A small note about the python cells.  In Jupyter notebooks, it works *something* like talking to the interpreter directly (>>>), and *mostly* like running the python as a script.

```python
# with only one line, it works like the REPL (>>>)
8+9
```

```python
# with more than one line, only the *last* line is treated like >>>
# all lines above are treated as a script
3+8
3**5
```

# f-strings and why they're awesome

```python
print('the result of 2313123 / 742 is' + 2313123 / 742)
```

```python
print('the result of 2313123 / 742 is ' + str(2313123 / 742))
```

```python
# this works but is very cumbersome:
print('the result of 2313123 / 742 is {}'.format(2313123 / 742))
```

```python
print('the result of 2313123 / 742 is {} and 2313123 / 543 is {}'.format(2313123 / 742, 2313123 / 543))
```

#### c'mon f-string!

```python
print(f'the result of 2313123 / 742 is {2313123 / 742} and 2313123 / 543 is {2313123 / 543}')
```

```python
# :2f is there to limit the number of decimals in the output
print(f'the result of 2313123 / 742 is {2313123 / 742:.2f} and 2313123 / 543 is {2313123 / 543:.2f}')
```

```python
x = 20

print('the value of x is ' + str(x))
print(f'the value of x is {x}')
print(f'{x}')
print(f'{x=}')
```
